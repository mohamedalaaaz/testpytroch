```python
def one_hot_encode(seq):
    mapping = {'A': 0, 'C': 1, 'G': 2, 'T': 3}
    encoding = torch.zeros(len(seq), 4)
    for i, nucleotide in enumerate(seq):
        if nucleotide in mapping:
            encoding[i, mapping[nucleotide]] = 1
    return encoding
```
